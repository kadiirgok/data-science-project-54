```python
import re
import zeyrek



import logging
logging.getLogger("zeyrek").setLevel(logging.ERROR)
analyzer = zeyrek.MorphAnalyzer()

def preprocess(text: str) -> str:
   
    # 1 Türkçe Stop-Word listesi
    standart_stopwords = {
        "ve", "veya", "da", "de", "ile", "için", "çünkü", "bir", "bu", "şu", "o",
        "ki", "mi", "mı", "mu", "mü", "gibi", "hem", "ise", "diye", "en", "daha",
        "böyle", "şöyle", "her", "hiç", "bazı", "tüm", "hep", "kendi",
        "ama", "fakat", "lakin", "ancak", "oysa", "halbuki", "yine", "rağmen"
    }
    
    #Silinmesini istemediğimiz zıtlık bağlaçları
    korunacak_zitlik_baglaclari = {
        "ama", "fakat", "lakin", "ancak", "oysa", "halbuki", "yine", "rağmen", "oysaki"
    }
    
    # Filtre  oluşturma
    temizlik_listesi = standart_stopwords - korunacak_zitlik_baglaclari

    #Türkçe Uyumlu Küçük Harfe Çevirme 
    harf_haritasi = {'İ': 'i', 'I': 'ı'}
    metin = text
    for buyuk, kucuk in harf_haritasi.items():
        metin = metin.replace(buyuk, kucuk)
    metin = metin.lower()

    #  Tekrar eden Harfleri Normalize 
    metin = re.sub(r'(\w)\1{2,}', r'\1', metin)

    # --- ADIM 3: Noktalama Temizliği (URL, E-posta ve Sayı Korumalı) ---
    # E-posta ve URL'leri korumak için etiketleme
    urls = re.findall(r'https?://\S+|www\.\S+', metin)
    emails = re.findall(r'\S+@\S+', metin)
    
    for i, url in enumerate(urls):
        metin = metin.replace(url, f" [URL_{i}] ")
    for i, email in enumerate(emails):
        metin = metin.replace(email, f" [EMAIL_{i}] ")

    # Kelime başı/sonu ve bağımsız noktalama işaretlerini temizleme
    metin = re.sub(r'(?<!\w)[.,!?;:"\'()\[\]{}–\-—<>‘“’”（）«»*_+~`|=#]+', ' ', metin)
    metin = re.sub(r'[.,!?;:"\'()\[\]{}–\-—<>‘“’”（）«»*_+~`|=#]+(?!\w)', ' ', metin)
    metin = re.sub(r'\s+[.,!?;:"\'()\[\]{}–\-—<>‘“’”（）«»*_+~`|=#]+\s+', ' ', metin)

    # E-posta ve URL'leri geri yükleme
    for i, url in enumerate(urls):
        metin = metin.replace(f" [URL_{i}] ", url)
    for i, email in enumerate(emails):
        metin = metin.replace(f" [EMAIL_{i}] ", email)

    # Fazla boşlukları temizleme
    metin = re.sub(r'\s+', ' ', metin).strip()

    #  Zeyrek ile Tokenize ve Lemmatization -
    analiz_sonuclari = analyzer.analyze(metin)
    
    kok_kelimeler = []
    
    for kelime_analizi in analiz_sonuclari:
     if not kelime_analizi:
        continue

     en_olasi_analiz = kelime_analizi[0]
     orijinal_kelime = en_olasi_analiz.word.lower()
     kelime_koku = getattr(en_olasi_analiz, 'lemma', en_olasi_analiz).lower()

    # Zıtlık bağlaçlarında zeyrek'e güvenme, orijinali kullan
     if orijinal_kelime in korunacak_zitlik_baglaclari:
        kelime_koku = orijinal_kelime

    # Zeyrek tanıyamadıysa (Unk), orijinali kullan
     if kelime_koku == "unk":
        kelime_koku = orijinal_kelime

     if not kelime_koku.isalpha():
        continue

     if kelime_koku not in temizlik_listesi:
        kok_kelimeler.append(kelime_koku)  
    # Temizlenmiş metni birleştirerek döndür
    processed_text = " ".join(kok_kelimeler)
    return processed_text
```

#Trian Test Split
klasik yöntem 80 20 ayrıdım datayı 

```python

from sklearn.model_selection import train_test_split

X = df["yorum_temiz"]
y = df["duygu"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(X_train.shape, X_test.shape)
print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

İlk yaklasım TF-IDF + Logistic Regression
`ngram_range=(1,2)` kullanıldı bazı duygu yapılarını daha iyi yaklaması için . `max_features=3000` ve `min_df=2` ile sözlüğü sınırlandırıldı.


```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(ngram_range=(1,2), max_features=3000, min_df=2)
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print(X_train_tfidf.shape, X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, classification_report

logreg = LogisticRegression(max_iter=1000, random_state=42)
logreg.fit(X_train_tfidf, y_train)

y_pred_train = logreg.predict(X_train_tfidf)
y_pred_test = logreg.predict(X_test_tfidf)

f1_train = f1_score(y_train, y_pred_train, pos_label="positive")
f1_test = f1_score(y_test, y_pred_test, pos_label="positive")

print(f"Train F1: {f1_train:.3f}")
print(f"Test F1: {f1_test:.3f}")
print()
print(classification_report(y_test, y_pred_test))
```

Train F1 = 0,937, Test F1 = 0,813. Aradaki fark (0,124) hafif bir overfitting'e işaret ediyor

```python
from sentence_transformers import SentenceTransformer

model_name = "emrecan/bert-base-turkish-cased-mean-nli-stsb-tr"
embed_model = SentenceTransformer(model_name)
```

İkinci yaklasım  Türkçe BERT Embedding + Logistic Regression
İkinci yaklaşım için embedding modeline hafif  bir temizlik uyguluyoruz — TF-IDF'teki ağır ön işlemeden (stopword silme, lemmatization) kasıtlı olarak farklı. Çünkü BERT tabanlı modeller cümlenin bağlamından (kelime sırası, ekler, noktalama) faydalanıyor.

```python
def hafif_temizle(text: str) -> str:
    metin = text.lower()  # Türkçe İ/ı sorunundan kaçınmak için istersen harf_haritasini burada da kullanabilirsin
    metin = re.sub(r'(\w)\1{2,}', r'\1', metin)  # tekrarlayan harfleri normalize et
    metin = re.sub(r'\s+', ' ', metin).strip()   # fazla boşlukları temizle
    return metin

df["yorum_hafif"] = df["yorum"].apply(hafif_temizle)
```

```python
X_train_hafif = df.loc[X_train.index, "yorum_hafif"]
X_test_hafif = df.loc[X_test.index, "yorum_hafif"]

X_train_emb = embed_model.encode(X_train_hafif.tolist(), show_progress_bar=True)
X_test_emb = embed_model.encode(X_test_hafif.tolist(), show_progress_bar=True)

print(X_train_emb.shape, X_test_emb.shape)
```

```python
logreg_emb = LogisticRegression(max_iter=1000, random_state=42)
logreg_emb.fit(X_train_emb, y_train)

y_pred_train_emb = logreg_emb.predict(X_train_emb)
y_pred_test_emb = logreg_emb.predict(X_test_emb)

f1_train_emb = f1_score(y_train, y_pred_train_emb, pos_label="positive")
f1_test_emb = f1_score(y_test, y_pred_test_emb, pos_label="positive")

print(f"Train F1 (embedding): {f1_train_emb:.3f}")
print(f"Test F1 (embedding): {f1_test_emb:.3f}")
print()
print(classification_report(y_test, y_pred_test_emb))
```

### İlk embedding sonucu 
Train F1 = 1,000, Test F1 = 0,847. Train F1'in tam 1,000 olması ciddi bir overfitting işareti — model train verisindeki her örneği ayırt edebiliyor, bu da gerçek bir örüntü öğrenmekten çok ezberlemiş demek oluyor


### C parametresini deniyoruz
Varsayılan ayarla (C=1.0) eğitilen embedding modelinde Train F1 = 1.000 çıktı — yani model train verisini ezberlemiş. Bunun düzelip düzelmediğini görmek için kontrol ediyorum


```python
#Model ezberleyip ezberlemediği kontrol ediyoruz
for C in [0.01, 0.1, 1.0]:
    m = LogisticRegression(max_iter=1000, random_state=42, C=C)
    m.fit(X_train_emb, y_train)
    f1_tr = f1_score(y_train, m.predict(X_train_emb), pos_label="positive")
    f1_te = f1_score(y_test, m.predict(X_test_emb), pos_label="positive")
    print(f"C={C}: Train F1={f1_tr:.3f}, Test F1={f1_te:.3f}")
```

### Regularization (C) denemesi sonucu
| C | Train F1 | Test F1 |
|---|---|---|
| 0,01 | 0,915 | **0,879** |
| 0,1 | 0,985 | 0,835 |
| 1,0 | 1,000 | 0,847 |

C=0,01'de hem overfitting neredeyse ortadan kalkıyor (train-test farkı 0,15'ten 0,04'e düşüyor) hem de Test F1 en yüksek seviyeye çıkıyor. Bu en iyi model oluyor


```python
logreg_emb_final = LogisticRegression(max_iter=1000, random_state=42, C=0.01)
logreg_emb_final.fit(X_train_emb, y_train)
```

```python
y_pred_final = logreg_emb_final.predict(X_test_emb)
print(classification_report(y_test, y_pred_final))
```

### Final embedding modeli sonucu
C=0,01 ile eğitilen final modelde precision/recall/f1 her iki sınıf için de dengeli çıktı 

**Genel karşılaştırma:**

| Yaklaşım | Test F1 |
|---|---|
| TF-IDF + Logistic Regression | 0,813 |
| Embedding + Logistic Regression (C=0,01) | **0,879** |


## 5. Hata Analizi
Bu bölümde test setindeki yanlış sınıflanan yorumları üç gruba ayırıp inceliyorum: sadece TF-IDF'in yanıldığı, sadece embedding'in yanıldığı, ve ikisinin de yanıldığı yorumlar var bunları göstermek


```python
# Test setindeki orijinal yorumları da görebilmek için bir dataframe 
test_df = df.loc[X_test.index, ["yorum", "duygu"]].copy()
test_df["tfidf_tahmin"] = logreg.predict(X_test_tfidf)
test_df["embedding_tahmin"] = logreg_emb_final.predict(X_test_emb)

# Her iki model de yanılmış mı, yoksa sadece biri mi?
test_df["tfidf_dogru"] = test_df["duygu"] == test_df["tfidf_tahmin"]
test_df["embedding_dogru"] = test_df["duygu"] == test_df["embedding_tahmin"]

print("Sadece TF-IDF yanıldı:", ((~test_df["tfidf_dogru"]) & (test_df["embedding_dogru"])).sum())
print("Sadece Embedding yanıldı:", ((test_df["tfidf_dogru"]) & (~test_df["embedding_dogru"])).sum())
print("İkisi de yanıldı:", ((~test_df["tfidf_dogru"]) & (~test_df["embedding_dogru"])).sum())
print("İkisi de doğru:", ((test_df["tfidf_dogru"]) & (test_df["embedding_dogru"])).sum())
```

### Karşılaştırma sayıları
Test setindeki 114 yorumun: 84'ünde iki model de doğru, 16'sında sadece TF-IDF yanıldı (embedding düzeltti), 7'sinde sadece embedding yanıldı, 7'sinde ikisi de yanıldı. 16 vs 7 oranı, embedding'in TF-IDF'e göre net bir üstünlüğü olduğunu gösteriyor — şimdi bu gruplardaki örnekleri tek tek inceleyip örüntü arayacağım.


```python
#Karsılastırma 
pd.set_option("display.max_colwidth", None)
ikisi_de_yanildi = test_df[(~test_df["tfidf_dogru"]) & (~test_df["embedding_dogru"])]
for _, row in ikisi_de_yanildi.iterrows():
    print(f"Gerçek: {row['duygu']} | TFIDF: {row['tfidf_tahmin']} | Embedding: {row['embedding_tahmin']}")
    print(row["yorum"])
    print()
```

### İkisi de yanılan örnekler
Bu grupta ilginç bir şey var: çoğu örnek aslında **açıkça pozitif okunan bir metin ama negative etiketlenmiş** (örn. "...tavsiye ediyorum. denerseniz neticeyi görürsünüz." → negative). Bu muhtemelen modelin hatası değil, **veri setindeki etiketleme gürültüsü** ya da kesik/eksik yorum metni — verinin kendi sınırlarını gösteren bir bulgu.


```python
#Tfıdf yanıldığı yerler 
sadece_tfidf_yanildi = test_df[(~test_df["tfidf_dogru"]) & (test_df["embedding_dogru"])]
for _, row in sadece_tfidf_yanildi.iterrows():
    print(f"Gerçek: {row['duygu']} | TFIDF tahmini: {row['tfidf_tahmin']}")
    print(row["yorum"])
    print()
```

```python
#Embeding yanıldığı yerler 
sadece_emb_yanildi = test_df[(test_df["tfidf_dogru"]) & (~test_df["embedding_dogru"])]
for _, row in sadece_emb_yanildi.iterrows():
    print(f"Gerçek: {row['duygu']} | Embedding tahmini: {row['embedding_tahmin']}")
    print(row["yorum"])
    print()
```

### Embedding'in yanıldığı yerler — zayıf yönü
Burada  yanıldığı örneklerin çoğu ters sıralı zıtlık cümleleri olumsuz başlayıp "ama/ancak" ile pozitife dönen cümleler. 

Bunun muhtemel sebebi: model, veride "ama/ancak"ın çoğunlukla negatife işaret ettiği genel örüntüyü çok iyi öğrenmiş (EDA'daki %69,5 bulgusu), ama bu örüntünün istisnalarında (nadir görülen ters sıralı cümlelerde) bu genellemeye fazla güvenip yanılıyor.

**Genel sonuç:** Embedding yöntemi, klasik yöntemin kaçırdığı bağlamsal ipuçlarını (negation, argo/küfür, cümle geneli) yakalayarak daha yüksek f1 elde ediyor, ama kendi öğrendiği istatistiksel örüntüye fazla bağlanabildiği durumlarda hataya düşüyor. Bu yüzden üretimde bu tip yorumları düşük güvenle işaretleyip insan gözden geçirmesine yönlendirmeli.
